```python
# Fundamental Booster Project
# Interactive Personal Data Collector

print("==========================================")
print(" Welcome to the Personal Data Collector ")
print("==========================================")

# Taking input from user
name = input("Enter your name: ")
age = int(input("Enter your age: "))
height = float(input("Enter your height in meters: "))
fav_number = int(input("Enter your favourite number: "))

print("\n===== Data Collected Successfully =====")

# Displaying values with type and memory address
print("\n--- User Information ---")

print("Name:", name)
print("Type of Name:", type(name))
print("Memory Address of Name:", id(name))

print("\nAge:", age)
print("Type of Age:", type(age))
print("Memory Address of Age:", id(age))

print("\nHeight:", height)
print("Type of Height:", type(height))
print("Memory Address of Height:", id(height))

print("\nFavourite Number:", fav_number)
print("Type of Favourite Number:", type(fav_number))
print("Memory Address of Favourite Number:", id(fav_number))

# Data Processing
birth_year = 2026 - age

print("\n===== Calculations =====")
print("Your approximate birth year is:", birth_year)

# Type Casting Example
height_int = int(height)

print("\n===== Type Casting =====")
print("Original Height:", height, "Type:", type(height))
print("Converted Height:", height_int, "Type:", type(height_int))

# Formatted Output
print("\n===== Summary =====")
print(f"Hello {name}!")
print(f"You are {age} years old.")
print(f"Your height is {height} meters.")
print(f"Your favourite number is {fav_number}.")
print(f"Your approximate birth year is {birth_year}.")

# Exit Message
print("\n==========================================")
print(" Thank you for using the program! ")
print(" Keep exploring Python Programming 🚀 ")
print("==========================================")
```
